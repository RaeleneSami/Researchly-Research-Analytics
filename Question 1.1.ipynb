```python

import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')


df = pd.read_json('research_abstracts.json')

print("Dataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst 5 Records:")
print(df.head())


print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Abstracts:",
      df.duplicated(subset=['Abstract']).sum())


df = df.dropna(subset=['Abstract', 'Label'])
df = df.drop_duplicates(subset=['Abstract']).copy()


stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()


def clean_text(text):

    text = str(text).lower()


    text = re.sub(r'http\S+|www\S+', '', text)

    
    text = re.sub(r'[^a-z\s]', ' ', text)


    tokens = word_tokenize(text)

    tokens = [
        stemmer.stem(word)
        for word in tokens
        if word not in stop_words and len(word) > 2
    ]

    return ' '.join(tokens)

df['Cleaned_Abstract'] = df['Abstract'].apply(clean_text)

df = df[df['Cleaned_Abstract'].str.strip() != ''].copy()

print("\nOriginal and Cleaned Abstracts:")
print(df[['Abstract', 'Cleaned_Abstract']].head())

print("\nFinal Dataset Shape:", df.shape)

df.to_csv('cleaned_research_abstracts.csv', index=False)

print("\nPreprocessing completed successfully!")
```
